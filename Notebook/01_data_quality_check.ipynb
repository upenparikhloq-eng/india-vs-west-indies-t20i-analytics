```python
import pandas as pd
import numpy as np
from pathlib import Path

# Project paths
PROJECT_DIR = Path(
    r"C:\India_Vs_West_Indies_T20i_Challenge"
)

RAW_DIR = PROJECT_DIR / "Data" / "Raw"
CLEAN_DIR = PROJECT_DIR / "Data" / "Clean"

CLEAN_DIR.mkdir(parents=True, exist_ok=True)

# Locate the CSV file
csv_files = list(RAW_DIR.glob("*.csv"))

if not csv_files:
    raise FileNotFoundError(
        f"No CSV file found in {RAW_DIR}"
    )

# Load the raw data
df = pd.read_csv(csv_files[0])

print("Dataset loaded successfully")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

df.head()
```

```python
# 1. Check dataset dimensions
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 2. Check column names and data types
df.info()

# 3. Display all column names
for col in df.columns:
    print(col)

# 4. Check first and last 5 records
display(df.head())
display(df.tail())
```

```python
# Missing values in each column
missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing_Count": missing_values,
    "Missing_Percentage": (
        missing_values / len(df) * 100
    ).round(2)
})

display(
    missing_report[
        missing_report["Missing_Count"] > 0
    ].sort_values(
        "Missing_Count",
        ascending=False
    )
)

# Exact duplicate rows
print("Exact duplicate rows:", df.duplicated().sum())

# Check completely empty rows
print("Completely empty rows:", df.isna().all(axis=1).sum())
```

```python
# Check numeric columns
display(df.describe(include="all").T)

# Check repeated player-name spellings
player_columns = [
    col for col in ["Batter", "Non-striker", "Bowler"]
    if col in df.columns
]

for col in player_columns:
    print(f"\nTop names in {col}:")
    display(df[col].value_counts(dropna=False).head(15))
```

```python
# Create a copy for auditing; keep the original df unchanged
audit_df = df.copy()

# Calculate expected runs from batter runs + extras
audit_df["Run Difference"] = (
    audit_df["Team Runs"]
    - audit_df["Runs Off The Bat"]
    - audit_df["Extra"]
)

# Compare Extra against the sum of recorded extras components
extra_columns = ["Wide", "No Ball", "Leg Bye", "Bye"]

audit_df["Extra Component Difference"] = (
    audit_df["Extra"]
    - audit_df[extra_columns].sum(axis=1)
)

# Expected four and six indicators based on runs off the bat
audit_df["Expected 4's"] = (
    audit_df["Runs Off The Bat"] == 4
).astype(int)

audit_df["Expected 6's"] = (
    audit_df["Runs Off The Bat"] == 6
).astype(int)

# Summary of flagged records
print("Team-run mismatches:",
      (audit_df["Run Difference"] != 0).sum())

print("Extras-component mismatches:",
      (audit_df["Extra Component Difference"] != 0).sum())

print("Incorrect four indicators:",
      (audit_df["4's"] != audit_df["Expected 4's"]).sum())

print("Incorrect six indicators:",
      (audit_df["6's"] != audit_df["Expected 6's"]).sum())
```

```python
# Display complete records for each issue
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

print("TEAM RUN MISMATCHES")
display(
    audit_df.loc[
        audit_df["Run Difference"] != 0,
        df.columns
    ]
)

print("EXTRAS COMPONENT MISMATCH")
display(
    audit_df.loc[
        audit_df["Extra Component Difference"] != 0,
        df.columns
    ]
)

print("INCORRECT FOUR INDICATORS")
display(
    audit_df.loc[
        audit_df["4's"] !=
        (audit_df["Runs Off The Bat"] == 4).astype(int),
        df.columns
    ]
)

# Check innings and team consistency
print("MATCH DAY AND INNINGS DISTRIBUTION")
display(
    df.groupby(
        ["MatchDay", "Innings", "Batting Team", "Bowling Team"],
        dropna=False
    ).agg(
        Deliveries=("Ball Number", "size"),
        Total_Runs=("Team Runs", "sum"),
        Batting_Runs=("Runs Off The Bat", "sum"),
        Extras=("Extra", "sum"),
        Wickets=("Wicket", "sum")
    ).reset_index()
)

# Check player-name spelling variants
for col in ["Striker", "Non-Striker", "Bowler"]:
    print(f"\nUnique names in {col}:")
    print(sorted(df[col].dropna().unique()))
```

```python
# Create a clean copy; do not modify the original df
clean_df = df.copy()

# 1. Standardize whitespace and obvious name variants
name_columns = ["Striker", "Non-Striker", "Bowler"]

for col in name_columns:
    clean_df[col] = (
        clean_df[col]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

name_mapping = {
    "NItish Reddy": "Nitish Reddy",
    "Prince Yaddav": "Prince Yadav"
}

for col in name_columns:
    clean_df[col] = clean_df[col].replace(name_mapping)

# 2. Recalculate boundary indicators from batter runs
clean_df["4's"] = (
    clean_df["Runs Off The Bat"] == 4
).astype(int)

clean_df["6's"] = (
    clean_df["Runs Off The Bat"] == 6
).astype(int)

# 3. Preserve unresolved run/extras issues for investigation
clean_df["Run_QC_Flag"] = (
    clean_df["Team Runs"]
    != clean_df["Runs Off The Bat"] + clean_df["Extra"]
)

clean_df["Extras_QC_Flag"] = (
    clean_df["Extra"]
    != clean_df[["Wide", "No Ball", "Leg Bye", "Bye"]].sum(axis=1)
)

# 4. Save the cleaned dataset
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

clean_path = CLEAN_DIR / "t20i_ball_by_ball_clean.csv"
clean_df.to_csv(clean_path, index=False)

print("Clean dataset saved:", clean_path)
print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
print("Run issues retained:", clean_df["Run_QC_Flag"].sum())
print("Extras issues retained:", clean_df["Extras_QC_Flag"].sum())
print("Four indicators now inconsistent:",
      (clean_df["4's"] !=
       (clean_df["Runs Off The Bat"] == 4).astype(int)).sum())
print("Six indicators now inconsistent:",
      (clean_df["6's"] !=
       (clean_df["Runs Off The Bat"] == 6).astype(int)).sum())
```

```python
match_summary = (
    clean_df.groupby(
        ["MatchDay", "Innings", "Batting Team", "Bowling Team"],
        as_index=False
    )
    .agg(
        Deliveries=("Ball Number", "size"),
        Total_Runs=("Team Runs", "sum"),
        Batter_Runs=("Runs Off The Bat", "sum"),
        Extras=("Extra", "sum"),
        Wickets=("Wicket", "sum"),
        Fours=("4's", "sum"),
        Sixes=("6's", "sum")
    )
)

display(match_summary)

# Check which matches have both innings recorded
print("\nInnings count per match:")
display(
    clean_df.groupby("MatchDay")["Innings"].nunique()
)

# Verify each innings has the expected batting/bowling team pairing
team_check = clean_df[
    ["MatchDay", "Innings", "Batting Team", "Bowling Team"]
].drop_duplicates()

display(
    team_check.sort_values(["MatchDay", "Innings"])
)
```

```python
# Rebuild the summary with explicit column names
innings_summary = (
    clean_df.groupby(
        ["MatchDay", "Innings", "Batting Team", "Bowling Team"],
        as_index=False
    )
    .agg(
        Deliveries=("Ball Number", "size"),
        Team_Runs=("Team Runs", "sum"),
        Batter_Runs=("Runs Off The Bat", "sum"),
        Extras=("Extra", "sum"),
        Wicket_Flags=("Wicket", "sum"),
        Fours=("4's", "sum"),
        Sixes=("6's", "sum")
    )
)

innings_summary["Runs_Reconciled"] = (
    innings_summary["Team_Runs"]
    == innings_summary["Batter_Runs"] + innings_summary["Extras"]
)

display(innings_summary.to_string(index=False))

# Inspect wicket values
print("\nWicket column values:")
print(clean_df["Wicket"].value_counts(dropna=False).sort_index())

# Show all rows marked as wickets
print("\nRows marked as wickets:")
display(
    clean_df.loc[
        clean_df["Wicket"] != 0,
        ["MatchDay", "Innings", "Striker", "Bowler",
         "Ball Number", "Team Runs", "Runs Off The Bat",
         "Extra", "Wicket"]
    ]
)
```

```python
# Investigate MatchDay 2, Innings 1
# and MatchDay 3, Innings 1

problem_innings = [(2, 1), (3, 1)]

for match_day, innings_no in problem_innings:
    print(f"\nMatchDay {match_day} | Innings {innings_no}")

    subset = clean_df[
        (clean_df["MatchDay"] == match_day)
        & (clean_df["Innings"] == innings_no)
    ].copy()

    subset["Calculated_Runs"] = (
        subset["Runs Off The Bat"] + subset["Extra"]
    )

    subset["Run_Difference"] = (
        subset["Team Runs"] - subset["Calculated_Runs"]
    )

    display(
        subset.loc[
            subset["Run_Difference"] != 0,
            [
                "Venue", "Striker", "Bowler",
                "Ball Number", "Actual Ball Number",
                "Team Runs", "Runs Off The Bat", "Extra",
                "Wide", "No Ball", "Leg Bye", "Bye",
                "Wicket", "Run_Difference"
            ]
        ]
    )

# Check invalid ball suffixes in the displayed ball number
ball_suffix = (
    clean_df["Ball Number"].mod(1).mul(10).round().astype(int)
)

print("\nBall-number suffix counts:")
display(ball_suffix.value_counts().sort_index())
```

```python
unusual_balls["Is_Wide"] = unusual_balls["Wide"].fillna(0) > 0
unusual_balls["Is_No_Ball"] = unusual_balls["No Ball"].fillna(0) > 0

display(
    unusual_balls.groupby(
        ["Is_Wide", "Is_No_Ball"],
        dropna=False
    ).size().reset_index(name="Delivery_Count")
)
```
